```python
# Write a program to implement the kNN classifier for a sample training data set stored as a .CSV file. Compute the accuracy of the classifier, considering few datasets. and also calculate accuracy, precision and recall for your dataset.

import pandas as pd
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv("titanic.csv")
df.head()
```

```python
# For checking which fields having how much null data

df.isna().sum()
```

```python
# Here we will need Survived, Pclass, Sex, Age, SibSp, Parch, Embarked for machine learning as feature or target
# But some of them data are not numerical
# and some columns are having null (empty) data

# So we will replace null data with median in numeric field and with mode in text field (Age and Embarked)
# Then we will convert text fields to numeric field (Sex and Embarked)

from numpy import median
df["Age"] = df["Age"].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

df['Sex'] = LabelEncoder().fit_transform(df['Sex'])
df['Embarked'] = LabelEncoder().fit_transform(df['Embarked'])

df.head()
```

```python
# Now we will divide feature and target (x and y)
x = df[['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Embarked']]
y = df['Survived']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()

x_train_scaler = scaler.fit_transform(x_train)
x_test_scaler = scaler.transform(x_test)
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train_scaler, y_train)
```

```python
y_pred = knn.predict(x_test_scaler)
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
```

```python
print(f'Accuracy:  {accuracy * 100}%')
print(f'Precision: {precision * 100}%')
print(f'Recall:    {recall * 100}%')
```
